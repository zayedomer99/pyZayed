filtering

```python
x=np.array([1,2,10,20])
print(x>=10)
print(x[x>=10])
print((x>=10) & (x<20))
print((x>=10) | (x<20))
```

```python
y=np.array([20,10,30,5])
x=np.where(y==10)
x1=np.where(y==10,"good","bad")
print(x)
print(y[x])
print(x1)
```

```python
y=np.array([50,60,100,20,80,30,45,99])
x=np.where(y<50,0,y)
print(x)
```
