```python
def gradedef(marks):
    if marks>80:
        return 'A'
    elif marks>60:
        return'B'
    elif marks>40:
        return'C'
    else:
        return'FAIL'

print(gradedef(40))
```

```python
students = pd.DataFrame({'student_id':[1,2,3],'name':['zayed','Ammar','kang',]})
marks = pd.DataFrame({'student_id':[1,2,3],'marks':[100,50,20]})
```

```python
pd.merge(
    students,
    marks,on="student_id",
    how="inner"
)
```

```python
students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','Adnan']})
marks = pd.DataFrame({'student_id':[1,2,3,4],'marks':[100,50,20,50]})
```

```python
pd.merge(
    students,
    marks,on="student_id",
    how="outer"
)
```

```python
pd.merge(
    students,
    marks,on="student_id",
    how="right"
)
```

```python
s=students = pd.DataFrame({'student_id':[1,2,3,6],'name':['Shahzeb','Ammar','Abdullah','Adnan']})
m=marks = pd.DataFrame({'id':[1,2,3,4],'marks':[100,50,20,50]})
```

```python
pd.merge(
    s,
    m,
    left_on='student_id',
    right_on='id',
    how='left'
)
```
